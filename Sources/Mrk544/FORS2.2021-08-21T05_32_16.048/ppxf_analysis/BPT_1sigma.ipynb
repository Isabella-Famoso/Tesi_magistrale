```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
spectrum_1 = pd.read_csv("Spectrum1_1_gas.csv")
spectrum_2 = pd.read_csv("Spectrum1_2_gas.csv")
spectrum_3 = pd.read_csv("Spectrum1_3_gas.csv")
spectrum_4 = pd.read_csv("Spectrum1_4_gas.csv")
```

```python
def Kauff_AGN(x):
    # AGN selection based on Kauffman et al. 2003
    # log OIII/Hb > 0.61/(log NII/Ha - 0.05) + 1.3
    return 0.61/(x - 0.05) + 1.3

def Kewley_AGN_NII(x):
     # AGN selection based on Kewley et al. 2001
     # log OIII/Hb > 0.61/(log NII/Ha - 0.47) + 1.19
     return 0.61/(x - 0.47) + 1.19
```

```python
plt.figure(figsize=(8,6))

OIII_HB_1 = spectrum_1['Flux'][7]/spectrum_1['Flux'][1]
NII_HA_1 = spectrum_1['Flux'][9]/spectrum_1['Flux'][2]

OIII_HB_2 = spectrum_2['Flux'][7]/spectrum_2['Flux'][1]
NII_HA_2 = spectrum_2['Flux'][9]/spectrum_2['Flux'][2]

OIII_HB_3 = spectrum_3['Flux'][7]/spectrum_3['Flux'][1]
NII_HA_3 = spectrum_3['Flux'][9]/spectrum_3['Flux'][2]

OIII_HB_4 = spectrum_4['Flux'][7]/spectrum_4['Flux'][1]
NII_HA_4 = spectrum_4['Flux'][9]/spectrum_4['Flux'][2]

plt.scatter(np.log(OIII_HB_1), np.log(NII_HA_1), color = 'blue')
plt.scatter(np.log(OIII_HB_2), np.log(NII_HA_2), color = 'green')
plt.scatter(np.log(OIII_HB_3), np.log(NII_HA_3), color = 'purple')
plt.scatter(np.log(OIII_HB_4), np.log(NII_HA_4), color = 'orange')

x = np.linspace(-1.5, 0.04, num=200)
x1 = np.linspace(-1.5, 0.457, num=200)
plt.plot(x, Kauff_AGN(x))
plt.plot(x1, Kewley_AGN_NII(x1))
plt.xlim(-1.5, 0.7)
plt.ylim(-1.5, 1.5)
plt.xlabel('log([NII]λ6583/Hα)', fontsize=12)
plt.ylabel('log([OIII]λ5007/Hβ)', fontsize=12)
```
